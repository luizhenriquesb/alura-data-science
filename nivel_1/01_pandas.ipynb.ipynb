## 1. Bibliotecas

```python
import pandas as pd
```

## 2. Importação

```python
df = pd.read_csv('https://raw.githubusercontent.com/alura-cursos/pandas-conhecendo-a-biblioteca/main/base-de-dados/aluguel.csv', sep=';')
```

## 3. Análise Exploratória

```python
df.info()
```

```python
# Média da coluna 'valor'
df.Valor.mean()
```

```python
# Média de valor por 'tipo'
(
    df
      .groupby('Tipo')
      .agg(media = ('Valor', 'mean'))
      .reset_index()
      .sort_values(by='media', ascending=False)
)
```
